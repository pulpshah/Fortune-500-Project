```python
from polygon import RESTClient
import datetime as dt
import pandas as pd
import time
```

```python
polygonAPIkey = 'YOUR-API-KEY

client = RESTClient(polygonAPIkey)
```

```python
import requests
from bs4 import BeautifulSoup


soup = BeautifulSoup(requests.get("https://stockanalysis.com/list/biggest-companies/").content, "html.parser")
```

```python
def getTickerTags(soup):

##The href tags contain the ticker tags for companies

    href_tags = soup.find_all(href=True)
    ticker_tags = []

    for tag in href_tags:
        if "/stocks/" in str(tag):
          ticker_tags.append(tag.text)


    #First four elements do not contain tags, so they are not included
    ticker_tags = ticker_tags[4:]

    return ticker_tags
```

```python
ticker_tags = getTickerTags(soup)
```

```python
financial_data = []
for tag in ticker_tags
    for t in client.vx.list_stock_financials(ticker=tag,
                                         filing_date_gte='2024-01-01'):
        financial_data.append(t)
        ##Time is required to avoid being rate limited
        time.sleep(15)
```

```python
df = pd.DataFrame(financial_data))
```

```python
def addRevenue(df):
    revenues = []

    for i, row in enumerate(df.iterrows()):
      if df["financials"][i]["income_statement"]["revenues"] == None:
          revenues.append(0)
      else:
        revenues.append(df["financials"][i]["income_statement"]["revenues"]["value"])

    return revenues
```

```python
df["revenues"] = revenues
```
